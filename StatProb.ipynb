```python
df = pd.read_csv('Metro_Interstate_Traffic_Volume.csv')
#memuat dataset
df['date_time'] = pd.to_datetime(df['date_time'], format = 'mixed')
#mengubah kolom date time dari string ke tipe date time, format ='mixed'
#agar dapat menebak da menyesuaikan format tanggal / waktu yang bervariasi
print(df.dtypes)
#nge print dataset yang sudah diubah
```

```python
kolom_list = {'traffic_volume', 'temp', 'clouds_all' }
#3 kolom yang akan dianalisis
ringkasan_data = []
#membuat fungsi atau ringkasan perhitungan untuk setiap data
for col in kolom_list:
    s = df[col]
    #ukuran pemusatan
    mean_val = s.mean()
    median_val = s.median()
    mode_val = s.mode()[0]
    #ukuran penyebaran
    min_val = s.min()
    max_val = s.max()
    jangkauan = max_val - min_val
    var_val = s.var()
    std_val = s.std()

    q1 = s.quantile(0.25)
    q3 = s.quantile(0.75)
    iqr = q3 - q1

    ringkasan_data.append({
        'Kolom': col,
        'Mean': mean_val,
        'Median': median_val,
        'Modus': mode_val,
        'Min': min_val,
        'Max': max_val,
        'Jangkauan': jangkauan,
        'Varians': var_val,
        'Simpangan Baku': std_val,
        'Q1': q1,
        'Q3': q3,
        'IQR': iqr
    
    })

tabel_ringkasan = pd.DataFrame(ringkasan_data)
display(tabel_ringkasan)
```
